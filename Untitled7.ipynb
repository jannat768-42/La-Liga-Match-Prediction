```python

# STEP 3: Create Match and MatchManager Classes
# Requirements: OOP, validation, functions, lists,
# searching, conditionals, loops, and exception handling

# ---------- Match Class ----------

class Match:
    def __init__(self, match_id, date, home_team, away_team,
                 home_goals, away_goals):

        self.match_id = match_id
        self.date = date
        self.home_team = home_team
        self.away_team = away_team
        self.home_goals = home_goals
        self.away_goals = away_goals

    # Determine the match result
    def get_result(self):
        if self.home_goals > self.away_goals:
            return "Home Win"
        elif self.home_goals < self.away_goals:
            return "Away Win"
        else:
            return "Draw"

    # Convert match object into a dictionary
    def to_dict(self):
        return {
            "match_id": self.match_id,
            "date": self.date,
            "home_team": self.home_team,
            "away_team": self.away_team,
            "home_goals": self.home_goals,
            "away_goals": self.away_goals
        }


# ---------- MatchManager Class ----------

class MatchManager:
    def __init__(self):
        self.matches = []
        self.next_id = 1

    # Add a new match with validation
    def add_match(self, date, home, away, home_goals, away_goals):
        date = str(date).strip()
        home = str(home).strip()
        away = str(away).strip()

        if not date or not home or not away:
            raise ValueError("Date and both team names are required.")

        try:
            match_date = datetime.strptime(date, "%Y-%m-%d")
        except ValueError:
            raise ValueError("Date must be in YYYY-MM-DD format.")

        if not 2021 <= match_date.year <= 2026:
            raise ValueError("Date must be between 2021 and 2026.")

        if home.casefold() == away.casefold():
            raise ValueError("Home and away teams must be different.")

        if type(home_goals) is not int or type(away_goals) is not int:
            raise ValueError("Goals must be integers.")

        if not (0 <= home_goals <= 20 and 0 <= away_goals <= 20):
            raise ValueError("Goals must be between 0 and 20.")

        # Prevent duplicate fixtures
        for match in self.matches:
            if (
                match.date == date
                and match.home_team.strip().casefold() == home.casefold()
                and match.away_team.strip().casefold() == away.casefold()
            ):
                raise ValueError("This match already exists.")

        new_match = Match(
            self.next_id, date, home, away,
            home_goals, away_goals
        )

        self.matches.append(new_match)
        self.next_id += 1
        return new_match

    # Search by match ID, date, or team name
    def search_matches(self, keyword):
        keyword = str(keyword).strip().casefold()

        if not keyword:
            raise ValueError("Search keyword cannot be empty.")

        return [
            match for match in self.matches
            if keyword in match.home_team.casefold()
            or keyword in match.away_team.casefold()
            or keyword in match.date
            or keyword == str(match.match_id)
        ]

    # Display all match records
    def view_matches(self):
        if not self.matches:
            print("No match records available.")
            return

        for match in self.matches:
            print(
                f"ID: {match.match_id} | "
                f"Date: {match.date} | "
                f"{match.home_team} "
                f"{match.home_goals}-{match.away_goals} "
                f"{match.away_team} | "
                f"Result: {match.get_result()}"
            )


# ---------- Load Sample CSV Records ----------

manager = MatchManager()

try:
    required_columns = ["Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG"]

    if not all(column in df.columns for column in required_columns):
        raise ValueError("CSV file is missing required columns.")

    for _, row in df.iterrows():
        try:
            date = pd.to_datetime(
                row["Date"], errors="raise"
            ).strftime("%Y-%m-%d")

            home_goals = int(row["FTHG"])
            away_goals = int(row["FTAG"])

            manager.add_match(
                date,
                str(row["HomeTeam"]),
                str(row["AwayTeam"]),
                home_goals,
                away_goals
            )

        except (ValueError, TypeError, OverflowError) as error:
            print("Skipped invalid record:", error)

    print("Total records loaded:", len(manager.matches))
    print("\nAll loaded matches:")
    manager.view_matches()

except NameError:
    print("Please run Step 2 first to create the sample DataFrame.")

except (ValueError, KeyError) as error:
    print("Could not load sample records:", error)
```

```python
# STEP 4: CRUD Operations
# Create, Read, Update, Delete, and Search match records

# 1. CREATE: Add a new match
def add_new_match():
    try:
        date = input("Enter date (YYYY-MM-DD): ").strip()
        home = input("Enter home team: ").strip()
        away = input("Enter away team: ").strip()

        home_goals = int(input("Enter home team goals: "))
        away_goals = int(input("Enter away team goals: "))

        match = manager.add_match(
            date, home, away, home_goals, away_goals
        )

        print("\nMatch added successfully!")
        print("Match ID:", match.match_id)
        print("Result:", match.get_result())

    except ValueError as error:
        print("Error:", error)


# 2. READ: View all matches
def view_all_matches():
    print("\n--- ALL MATCHES ---")
    manager.view_matches()


# 3. SEARCH: Find a match by ID, date, or team name
def search_match():
    try:
        keyword = input(
            "Enter Match ID, date, or team name: "
        ).strip()

        results = manager.search_matches(keyword)

        if not results:
            print("No matching match found.")
            return

        print("\n--- SEARCH RESULTS ---")
        for match in results:
            print(
                f"ID: {match.match_id} | "
                f"{match.date} | "
                f"{match.home_team} "
                f"{match.home_goals}-{match.away_goals} "
                f"{match.away_team} | "
                f"{match.get_result()}"
            )

    except ValueError as error:
        print("Error:", error)


# 4. UPDATE: Change a match's score
def update_match():
    try:
        match_id = int(input("Enter Match ID to update: "))

        match = next(
            (m for m in manager.matches if m.match_id == match_id),
            None
        )

        if match is None:
            print("Match ID not found.")
            return

        print(
            "\nCurrent match:",
            match.home_team,
            f"{match.home_goals}-{match.away_goals}",
            match.away_team
        )

        home_goals = int(input("Enter new home goals (0-20): "))
        away_goals = int(input("Enter new away goals (0-20): "))

        if not (0 <= home_goals <= 20 and 0 <= away_goals <= 20):
            print("Goals must be between 0 and 20.")
            return

        # Update only after both scores pass validation
        match.home_goals = home_goals
        match.away_goals = away_goals

        print("Match updated successfully!")
        print("New result:", match.get_result())

    except ValueError:
        print("Please enter a valid numeric Match ID and goals.")


# 5. DELETE: Remove a match by ID
def delete_match():
    try:
        match_id = int(input("Enter Match ID to delete: "))

        match = next(
            (m for m in manager.matches if m.match_id == match_id),
            None
        )

        if match is None:
            print("Match ID not found.")
            return

        manager.matches.remove(match)
        print("Match deleted successfully!")

    except ValueError:
        print("Please enter a valid numeric Match ID.")


print("Step 4: All match management functions are ready!")
```

```python

# STEP 5: Save and Load Match Records using JSON

DATA_FILE = "la_liga_records.json"


# Save all match records
def save_matches():
    try:
        records = [match.to_dict() for match in manager.matches]

        with open(DATA_FILE, "w", encoding="utf-8") as file:
            json.dump(records, file, indent=4)

        print("Matches saved successfully!")
        print("Total records saved:", len(records))

    except (OSError, TypeError) as error:
        print("Could not save records:", error)


# Load and validate match records
def load_matches():
    try:
        with open(DATA_FILE, "r", encoding="utf-8") as file:
            records = json.load(file)

        if not isinstance(records, list):
            raise ValueError("Expected a list of match records.")

        loaded_matches = []
        used_ids = set()
        used_fixtures = set()

        for record in records:
            try:
                if not isinstance(record, dict):
                    raise ValueError("Record must be a dictionary.")

                match_id = record["match_id"]
                date = record["date"]
                home = record["home_team"]
                away = record["away_team"]
                home_goals = record["home_goals"]
                away_goals = record["away_goals"]

                # Validate data types
                if type(match_id) is not int or match_id < 1:
                    raise ValueError("Match ID must be a positive integer.")

                if not all(
                    isinstance(value, str) and value.strip()
                    for value in [date, home, away]
                ):
                    raise ValueError("Date and team names are required.")

                if type(home_goals) is not int or type(away_goals) is not int:
                    raise ValueError("Goals must be integers.")

                # Validate date
                parsed_date = datetime.strptime(date, "%Y-%m-%d")

                if not 2021 <= parsed_date.year <= 2026:
                    raise ValueError("Year must be between 2021 and 2026.")

                home = home.strip()
                away = away.strip()

                if home.casefold() == away.casefold():
                    raise ValueError("Home and away teams must differ.")

                if not (0 <= home_goals <= 20 and 0 <= away_goals <= 20):
                    raise ValueError("Goals must be between 0 and 20.")

                # Detect duplicate IDs
                if match_id in used_ids:
                    raise ValueError("Duplicate match ID.")

                # Detect duplicate fixtures
                fixture = (
                    date,
                    home.casefold(),
                    away.casefold()
                )

                if fixture in used_fixtures:
                    raise ValueError("Duplicate match fixture.")

                # Create Match object
                match = Match(
                    match_id, date, home, away,
                    home_goals, away_goals
                )

                loaded_matches.append(match)
                used_ids.add(match_id)
                used_fixtures.add(fixture)

            except (ValueError, TypeError, KeyError) as error:
                print("Skipped invalid record:", error)

        # Replace in-memory records after loading
        manager.matches = loaded_matches
        manager.next_id = max(used_ids, default=0) + 1

        print("Matches loaded successfully!")
        print("Total valid records:", len(manager.matches))

    except FileNotFoundError:
        print("No saved file found. Save matches first.")

    except json.JSONDecodeError:
        print("The JSON file is corrupted or contains invalid JSON.")

    except (OSError, ValueError) as error:
        print("Could not load records:", error)


print("Step 5: Save and Load functions created successfully!")
```

```python

# STEP 6: Team Statistics

def team_statistics():
    team_name = input("Enter team name: ").strip()

    if not team_name:
        print("Team name cannot be empty.")
        return

    played = 0
    wins = 0
    draws = 0
    losses = 0
    goals_for = 0
    goals_against = 0

    for match in manager.matches:
        if match.home_team.casefold() == team_name.casefold():
            played += 1
            goals_for += match.home_goals
            goals_against += match.away_goals

            if match.home_goals > match.away_goals:
                wins += 1
            elif match.home_goals == match.away_goals:
                draws += 1
            else:
                losses += 1

        elif match.away_team.casefold() == team_name.casefold():
            played += 1
            goals_for += match.away_goals
            goals_against += match.home_goals

            if match.away_goals > match.home_goals:
                wins += 1
            elif match.away_goals == match.home_goals:
                draws += 1
            else:
                losses += 1

    if played == 0:
        print("No match records found for this team.")
        return

    points = wins * 3 + draws
    goal_difference = goals_for - goals_against
    win_percentage = (wins / played) * 100

    print("\n--- TEAM STATISTICS ---")
    print("Team:", team_name)
    print("Matches Played:", played)
    print("Wins:", wins)
    print("Draws:", draws)
    print("Losses:", losses)
    print("Goals Scored:", goals_for)
    print("Goals Conceded:", goals_against)
    print("Goal Difference:", goal_difference)
    print("League Points:", points)
    print("Win Percentage:", round(win_percentage, 2), "%")


print("Team Statistics function created!")
```

```python
# STEP 6.2: Baseline Match Predictor
# Uses historical average goals; this is not trained ML.

def predict_match():
    if not manager.matches:
        print("No match records available for prediction.")
        return

    home = input("Enter home team: ").strip()
    away = input("Enter away team: ").strip()

    if not home or not away:
        print("Both team names are required.")
        return

    if home.casefold() == away.casefold():
        print("Please enter two different teams.")
        return

    home_matches = [
        m for m in manager.matches
        if m.home_team.casefold() == home.casefold()
        or m.away_team.casefold() == home.casefold()
    ]

    away_matches = [
        m for m in manager.matches
        if m.home_team.casefold() == away.casefold()
        or m.away_team.casefold() == away.casefold()
    ]

    if not home_matches or not away_matches:
        print("Not enough historical data for one or both teams.")
        return

    def average_goals(team, matches, scoring=True):
        goals = []

        for match in matches:
            is_home = match.home_team.casefold() == team.casefold()

            if is_home:
                goals.append(
                    match.home_goals if scoring else match.away_goals
                )
            else:
                goals.append(
                    match.away_goals if scoring else match.home_goals
                )

        return sum(goals) / len(goals)

    home_attack = average_goals(home, home_matches, scoring=True)
    home_defence = average_goals(home, home_matches, scoring=False)

    away_attack = average_goals(away, away_matches, scoring=True)
    away_defence = average_goals(away, away_matches, scoring=False)

    predicted_home = (home_attack + away_defence) / 2
    predicted_away = (away_attack + home_defence) / 2

    print("\n--- BASELINE MATCH PREDICTION ---")
    print(f"Home Team: {home}")
    print(f"Away Team: {away}")
    print(f"Predicted home goals: {predicted_home:.2f}")
    print(f"Predicted away goals: {predicted_away:.2f}")

    if predicted_home > predicted_away:
        print("Predicted outcome: Home Win")
    elif predicted_home < predicted_away:
        print("Predicted outcome: Away Win")
    else:
        print("Predicted outcome: Draw")

    print(
        "\nNote: This is an average-goals baseline, "
        "not a trained Machine Learning model."
    )


print("Match Predictor function created!")
```

```python

# STEP 7: Fixed La Liga Dashboard for Google Colab

from IPython.display import display, clear_output
import ipywidgets as widgets

output_area = widgets.Output()
form_area = widgets.VBox()


def show_message(message):
    with output_area:
        clear_output(wait=True)
        print(message)


def show_home(_=None):
    form_area.children = ()
    show_message(
        "LA LIGA FOOTBALL MATCH PREDICTOR\n"
        "---------------------------------\n"
        "Welcome!\n"
        f"Total matches: {len(manager.matches)}\n"
        "Choose an operation using the buttons."
    )


# ---------- ADD MATCH ----------

date_box = widgets.Text(description="Date:", placeholder="YYYY-MM-DD")
home_box = widgets.Text(description="Home:")
away_box = widgets.Text(description="Away:")

hg_box = widgets.BoundedIntText(
    description="Home goals:", value=0, min=0, max=20
)
ag_box = widgets.BoundedIntText(
    description="Away goals:", value=0, min=0, max=20
)


def add_page(_):
    submit = widgets.Button(
        description="Submit Match", button_style="success"
    )

    def submit_action(_):
        with output_area:
            clear_output(wait=True)
            try:
                m = manager.add_match(
                    date_box.value.strip(),
                    home_box.value.strip(),
                    away_box.value.strip(),
                    hg_box.value,
                    ag_box.value
                )
                print("Match added successfully!")
                print("Match ID:", m.match_id)
                print("Result:", m.get_result())
                print("Total matches:", len(manager.matches))

            except (ValueError, TypeError) as e:
                print("Error:", e)

    submit.on_click(submit_action)

    form_area.children = (
        widgets.HTML("<h3>Add New Match</h3>"),
        date_box, home_box, away_box, hg_box, ag_box, submit
    )
    show_message("Fill in the fields, then click Submit Match.")


# ---------- VIEW MATCHES ----------

def view_page(_):
    form_area.children = ()

    with output_area:
        clear_output(wait=True)
        print("ALL MATCHES")
        print("-" * 50)

        if not manager.matches:
            print("No matches available.")
        else:
            for m in manager.matches:
                print(
                    f"ID: {m.match_id} | {m.date} | "
                    f"{m.home_team} {m.home_goals}-{m.away_goals} "
                    f"{m.away_team} | {m.get_result()}"
                )


# ---------- SEARCH ----------

search_box = widgets.Text(
    description="Search:", placeholder="Team, date or match ID"
)


def search_page(_):
    submit = widgets.Button(
        description="Search", button_style="info"
    )

    def do_search(_):
        with output_area:
            clear_output(wait=True)

            try:
                results = manager.search_matches(search_box.value)

                if not results:
                    print("No matching match found.")
                    return

                print("SEARCH RESULTS")
                print("-" * 50)

                for m in results:
                    print(
                        f"ID: {m.match_id} | {m.date} | "
                        f"{m.home_team} {m.home_goals}-{m.away_goals} "
                        f"{m.away_team} | {m.get_result()}"
                    )

            except ValueError as e:
                print("Error:", e)

    submit.on_click(do_search)

    form_area.children = (
        widgets.HTML("<h3>Search Matches</h3>"),
        search_box, submit
    )
    show_message("Enter a team name, date or Match ID.")


# ---------- UPDATE ----------

id_update = widgets.BoundedIntText(
    description="Match ID:", value=1, min=1
)

new_hg = widgets.BoundedIntText(
    description="New home goals:", value=0, min=0, max=20
)

new_ag = widgets.BoundedIntText(
    description="New away goals:", value=0, min=0, max=20
)


def update_page(_):
    submit = widgets.Button(
        description="Update Match", button_style="warning"
    )

    def do_update(_):
        with output_area:
            clear_output(wait=True)

            m = next(
                (x for x in manager.matches
                 if x.match_id == id_update.value),
                None
            )

            if m is None:
                print("Match ID not found.")
                return

            m.home_goals = new_hg.value
            m.away_goals = new_ag.value

            print("Match updated successfully!")
            print(
                f"{m.home_team} {m.home_goals}-"
                f"{m.away_goals} {m.away_team}"
            )
            print("New result:", m.get_result())
            print("Remember to save changes using Save Matches.")

    submit.on_click(do_update)

    form_area.children = (
        widgets.HTML("<h3>Update Match</h3>"),
        id_update, new_hg, new_ag, submit
    )
    show_message("Enter the Match ID and new goal values.")


# ---------- DELETE ----------

id_delete = widgets.BoundedIntText(
    description="Match ID:", value=1, min=1
)


def delete_page(_):
    submit = widgets.Button(
        description="Delete Match", button_style="danger"
    )

    def do_delete(_):
        with output_area:
            clear_output(wait=True)

            m = next(
                (x for x in manager.matches
                 if x.match_id == id_delete.value),
                None
            )

            if m is None:
                print("Match ID not found.")
                return

            manager.matches.remove(m)
            print("Match deleted successfully!")
            print("Total matches:", len(manager.matches))
            print("Remember to save changes using Save Matches.")

    submit.on_click(do_delete)

    form_area.children = (
        widgets.HTML("<h3>Delete Match</h3>"),
        id_delete, submit
    )
    show_message("Enter the Match ID to delete.")


# ---------- STATISTICS ----------

team_box = widgets.Text(description="Team:")


def stats_page(_):
    submit = widgets.Button(description="Show Statistics")

    def do_stats(_):
        team = team_box.value.strip().casefold()

        with output_area:
            clear_output(wait=True)

            if not team:
                print("Please enter a team name.")
                return

            played = wins = draws = losses = 0
            gf = ga = 0

            for m in manager.matches:
                if m.home_team.casefold() == team:
                    played += 1
                    gf += m.home_goals
                    ga += m.away_goals

                    if m.home_goals > m.away_goals:
                        wins += 1
                    elif m.home_goals == m.away_goals:
                        draws += 1
                    else:
                        losses += 1

                elif m.away_team.casefold() == team:
                    played += 1
                    gf += m.away_goals
                    ga += m.home_goals

                    if m.away_goals > m.home_goals:
                        wins += 1
                    elif m.away_goals == m.home_goals:
                        draws += 1
                    else:
                        losses += 1

            if played == 0:
                print("No records found for this team.")
                return

            points = wins * 3 + draws
            goal_difference = gf - ga
            win_percentage = (wins / played) * 100

            print("TEAM STATISTICS")
            print("Team:", team_box.value.strip())
            print("Matches Played:", played)
            print("Wins:", wins)
            print("Draws:", draws)
            print("Losses:", losses)
            print("Goals For:", gf)
            print("Goals Against:", ga)
            print("Goal Difference:", goal_difference)
            print("League Points:", points)
            print("Win Percentage:", round(win_percentage, 2), "%")

    submit.on_click(do_stats)

    form_area.children = (
        widgets.HTML("<h3>Team Statistics</h3>"),
        team_box, submit
    )
    show_message("Enter a team name to calculate statistics.")


# ---------- PREDICTOR ----------

pred_home = widgets.Text(description="Home team:")
pred_away = widgets.Text(description="Away team:")


def predict_page(_):
    submit = widgets.Button(
        description="Predict Match", button_style="info"
    )

    def do_predict(_):
        h = pred_home.value.strip()
        a = pred_away.value.strip()

        with output_area:
            clear_output(wait=True)

            if not manager.matches:
                print("No match records available for prediction.")
                return

            if not h or not a:
                print("Both team names are required.")
                return

            if h.casefold() == a.casefold():
                print("Enter two different team names.")
                return

            def averages(team):
                scored = []
                conceded = []
                team_key = team.casefold()

                for m in manager.matches:
                    if m.home_team.casefold() == team_key:
                        scored.append(m.home_goals)
                        conceded.append(m.away_goals)

                    elif m.away_team.casefold() == team_key:
                        scored.append(m.away_goals)
                        conceded.append(m.home_goals)

                if not scored:
                    return None

                return (
                    sum(scored) / len(scored),
                    sum(conceded) / len(conceded)
                )

            hs = averages(h)
            aws = averages(a)

            if hs is None or aws is None:
                print(
                    "Not enough match records for one or both teams."
                )
                return

            home_score = (hs[0] + aws[1]) / 2
            away_score = (aws[0] + hs[1]) / 2

            print("BASELINE PREDICTION")
            print("(Not a trained Machine Learning model)")
            print(f"{h}: {home_score:.2f} estimated goals")
            print(f"{a}: {away_score:.2f} estimated goals")

            if home_score > away_score:
                print("Predicted outcome: Home Win")
            elif away_score > home_score:
                print("Predicted outcome: Away Win")
            else:
                print("Predicted outcome: Draw")

    submit.on_click(do_predict)

    form_area.children = (
        widgets.HTML("<h3>Match Predictor</h3>"),
        pred_home, pred_away, submit
    )
    show_message("Enter the two teams.")


# ---------- SAVE / LOAD ----------

def save_page(_):
    form_area.children = ()

    with output_area:
        clear_output(wait=True)
        save_matches()


def load_page(_):
    form_area.children = ()

    with output_area:
        clear_output(wait=True)
        load_matches()
        print("\nCurrent total matches:", len(manager.matches))


# ---------- BUTTONS ----------

buttons = [
    ("Home", show_home),
    ("Add Match", add_page),
    ("View Matches", view_page),
    ("Search", search_page),
    ("Update", update_page),
    ("Delete", delete_page),
    ("Statistics", stats_page),
    ("Predictor", predict_page),
    ("Save Matches", save_page),
    ("Load Matches", load_page)
]

button_widgets = []

for label, callback in buttons:
    b = widgets.Button(description=label)
    b.on_click(callback)
    button_widgets.append(b)

row1 = widgets.HBox(button_widgets[:4])
row2 = widgets.HBox(button_widgets[4:8])
row3 = widgets.HBox(button_widgets[8:])

dashboard = widgets.VBox([
    widgets.HTML("<h2>La Liga Football Match Predictor</h2>"),
    row1, row2, row3,
    form_area,
    output_area
])

display(dashboard)
show_home()
```

```python
# STEP 8: Tuples, Sets and Data Validation
# 1. Tuple: Store the project period
PROJECT_YEARS = (2021, 2022, 2023, 2024, 2025, 2026)

print("Project years:", PROJECT_YEARS)


# 2. Set: Store unique team names
def get_unique_teams():
    teams = set()

    for match in manager.matches:
        teams.add(match.home_team.strip())
        teams.add(match.away_team.strip())

    return teams


unique_teams = get_unique_teams()

print("\nUnique teams:", unique_teams)
print("Total unique teams:", len(unique_teams))


# 3. Dictionary: Count match outcomes
def count_match_results():
    results = {
        "Home Win": 0,
        "Away Win": 0,
        "Draw": 0
    }

    for match in manager.matches:
        results[match.get_result()] += 1

    return results


print("\nMatch result summary:")
print(count_match_results())


# 4. Check for invalid or duplicate records
def validate_records():
    seen = set()
    invalid = 0

    for match in manager.matches:
        date = str(match.date).strip()
        home = str(match.home_team).strip()
        away = str(match.away_team).strip()

        key = (
            date,
            home.casefold(),
            away.casefold()
        )

        is_invalid = (
            not date
            or not home
            or not away
            or home.casefold() == away.casefold()
            or type(match.home_goals) is not int
            or type(match.away_goals) is not int
            or not (0 <= match.home_goals <= 20)
            or not (0 <= match.away_goals <= 20)
            or key in seen
        )

        if is_invalid:
            invalid += 1

        seen.add(key)

    print("\nInvalid or duplicate records:", invalid)
    print("Validation check completed.")

    return invalid


validate_records()
```
