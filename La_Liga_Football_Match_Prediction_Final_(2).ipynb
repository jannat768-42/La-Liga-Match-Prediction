```python
# PART 5: MATCH RESULT CALCULATION FUNCTIONS

def calculate_match_result(home_goals, away_goals):

    if home_goals > away_goals:
        return "Home Win"

    elif away_goals > home_goals:
        return "Away Win"

    else:
        return "Draw"


def calculate_team_points(home_goals, away_goals):

    if home_goals > away_goals:
        return 3, 0

    elif away_goals > home_goals:
        return 0, 3

    else:
        return 1, 1


# Calculate results for all matches
for match in matches:

    match["result"] = calculate_match_result(
        match["home_goals"],
        match["away_goals"]
    )

    home_points, away_points = calculate_team_points(
        match["home_goals"],
        match["away_goals"]
    )

    match["home_points"] = home_points
    match["away_points"] = away_points


# Update DataFrame
matches_df = pd.DataFrame(matches)

print("MATCH RESULT CALCULATIONS")

display(
    matches_df[
        [
            "match_id",
            "home_team",
            "away_team",
            "home_goals",
            "away_goals",
            "result",
            "home_points",
            "away_points"
        ]
    ]
)

print("\nMatch results calculated successfully!")
```

```python
# PART 6: OBJECT-ORIENTED PROGRAMMING

class Match:

    def __init__(
        self,
        match_id,
        date,
        home_team,
        away_team,
        home_goals,
        away_goals
    ):

        self.match_id = match_id
        self.date = date
        self.home_team = home_team
        self.away_team = away_team
        self.home_goals = home_goals
        self.away_goals = away_goals

    def get_result(self):

        if self.home_goals > self.away_goals:
            return "Home Win"

        elif self.away_goals > self.home_goals:
            return "Away Win"

        else:
            return "Draw"

    def get_points(self):

        if self.home_goals > self.away_goals:
            return 3, 0

        elif self.away_goals > self.home_goals:
            return 0, 3

        else:
            return 1, 1

    def display_match(self):

        print("Match ID:", self.match_id)
        print("Date:", self.date)
        print("Home Team:", self.home_team)
        print("Away Team:", self.away_team)
        print(
            "Score:",
            self.home_goals,
            "-",
            self.away_goals
        )
        print("Result:", self.get_result())


# Create Match Objects
match_objects = []

for match in matches:

    match_object = Match(
        match["match_id"],
        match["date"],
        match["home_team"],
        match["away_team"],
        match["home_goals"],
        match["away_goals"]
    )

    match_objects.append(match_object)


# Display first match
print("OOP MATCH OBJECT DEMONSTRATION")

match_objects[0].display_match()

print("\nTotal Match Objects:", len(match_objects))
```

```python
# PART 7: INPUT VALIDATION FUNCTIONS

def validate_team_name(team_name):

    if not isinstance(team_name, str):
        raise ValueError("Team name must be text.")

    team_name = team_name.strip()

    if team_name == "":
        raise ValueError("Team name cannot be empty.")

    if team_name not in TEAM_NAMES:
        raise ValueError("Team not found in the team list.")

    return team_name


def validate_goals(goals):

    try:
        goals = int(goals)

    except (ValueError, TypeError):
        raise ValueError("Goals must be a valid number.")

    if goals < 0:
        raise ValueError("Goals cannot be negative.")

    if goals > 20:
        raise ValueError("Goals cannot be greater than 20.")

    return goals


def validate_match_teams(home_team, away_team):

    home_team = validate_team_name(home_team)
    away_team = validate_team_name(away_team)

    if home_team == away_team:
        raise ValueError("Home and Away teams cannot be the same.")

    return home_team, away_team

# TEST VALIDATION FUNCTIONS

print("INPUT VALIDATION TEST")

try:

    home = validate_team_name("Real Madrid")
    away = validate_team_name("Barcelona")

    home_goals = validate_goals("2")
    away_goals = validate_goals("1")

    home, away = validate_match_teams(home, away)

    print("Home Team:", home)
    print("Away Team:", away)
    print("Home Goals:", home_goals)
    print("Away Goals:", away_goals)

    print("\nValidation Successful!")

except ValueError as error:

    print("Validation Error:", error)
```

```python
# PART 8: MATCH MANAGER CLASS

class MatchManager:

    def __init__(self, match_records):

        self.matches = match_records

    # 1. Add a new match
    def add_match(
        self,
        match_id,
        date,
        home_team,
        away_team,
        home_goals,
        away_goals
    ):

        match_id = str(match_id).strip()
        date = str(date).strip()

        if match_id == "":
            raise ValueError("Match ID cannot be empty.")

        if date == "":
            raise ValueError("Match date cannot be empty.")

        # Check duplicate ID
        for match in self.matches:

            if match["match_id"] == match_id:
                raise ValueError("Duplicate Match ID!")

        # Validate teams and goals
        home_team, away_team = validate_match_teams(
            home_team,
            away_team
        )

        home_goals = validate_goals(home_goals)
        away_goals = validate_goals(away_goals)

        # Calculate result and points
        result = calculate_match_result(
            home_goals,
            away_goals
        )

        home_points, away_points = calculate_team_points(
            home_goals,
            away_goals
        )

        # Create new match dictionary
        new_match = {
            "match_id": match_id,
            "date": date,
            "home_team": home_team,
            "away_team": away_team,
            "home_goals": home_goals,
            "away_goals": away_goals,
            "result": result,
            "home_points": home_points,
            "away_points": away_points
        }

        self.matches.append(new_match)

        return "Match added successfully!"

    # 2. View all matches
    def view_matches(self):

        return self.matches

    # 3. Search match by ID
    def search_match(self, match_id):

        for match in self.matches:

            if match["match_id"] == str(match_id).strip():
                return match

        return None

    # 4. Update match score
    def update_match(
        self,
        match_id,
        home_goals,
        away_goals
    ):

        match = self.search_match(match_id)

        if match is None:
            raise ValueError("Match not found!")

        home_goals = validate_goals(home_goals)
        away_goals = validate_goals(away_goals)

        match["home_goals"] = home_goals
        match["away_goals"] = away_goals

        match["result"] = calculate_match_result(
            home_goals,
            away_goals
        )

        home_points, away_points = calculate_team_points(
            home_goals,
            away_goals
        )

        match["home_points"] = home_points
        match["away_points"] = away_points

        return "Match updated successfully!"

    # 5. Delete match
    def delete_match(self, match_id):

        match = self.search_match(match_id)

        if match is None:
            raise ValueError("Match not found!")

        self.matches.remove(match)

        return "Match deleted successfully!"


# Create Match Manager Object
match_manager = MatchManager(matches)


# TEST MATCH MANAGER

print("MATCH MANAGER TEST")

# View total matches
print("Total Matches:", len(match_manager.view_matches()))

# Search a match
searched_match = match_manager.search_match("M001")

if searched_match is not None:

    print("\nMatch Found:")
    print("Match ID:", searched_match["match_id"])
    print("Home Team:", searched_match["home_team"])
    print("Away Team:", searched_match["away_team"])
    print(
        "Score:",
        searched_match["home_goals"],
        "-",
        searched_match["away_goals"]
    )

else:
    print("Match not found!")
```

```python
# PART 9: FILE HANDLING AND DATA PERSISTENCE

# Create a folder for project files
DATA_FOLDER = "la_liga_data"

os.makedirs(DATA_FOLDER, exist_ok=True)

# File paths
JSON_FILE = os.path.join(DATA_FOLDER, "matches.json")
CSV_FILE = os.path.join(DATA_FOLDER, "matches.csv")
TXT_FILE = os.path.join(DATA_FOLDER, "matches.txt")


# 1. Save matches as JSON
def save_matches_to_json(match_records):

    try:
        with open(JSON_FILE, "w") as file:

            json.dump(
                match_records,
                file,
                indent=4
            )

        print("JSON file saved successfully!")

    except (OSError, TypeError) as error:

        print("JSON Save Error:", error)


# 2. Load matches from JSON
def load_matches_from_json():

    try:
        with open(JSON_FILE, "r") as file:

            loaded_data = json.load(file)

        if not isinstance(loaded_data, list):
            raise ValueError("Invalid match data format.")

        for match in loaded_data:

            required_fields = [
                "match_id",
                "date",
                "home_team",
                "away_team",
                "home_goals",
                "away_goals"
            ]

            if not isinstance(match, dict):
                raise ValueError("Invalid match record.")

            for field in required_fields:

                if field not in match:
                    raise ValueError("Missing field: " + field)

        return loaded_data

    except FileNotFoundError:

        print("JSON file not found!")
        return None

    except (json.JSONDecodeError, ValueError) as error:

        print("JSON Data Error:", error)
        return None

    except OSError as error:

        print("JSON Load Error:", error)
        return None


# 3. Save matches as CSV
def save_matches_to_csv(match_records):

    try:

        if not match_records:
            print("No match records to save.")
            return

        fieldnames = list(match_records[0].keys())

        with open(CSV_FILE, "w", newline="") as file:

            writer = csv.DictWriter(
                file,
                fieldnames=fieldnames
            )

            writer.writeheader()
            writer.writerows(match_records)

        print("CSV file saved successfully!")

    except OSError as error:

        print("CSV Save Error:", error)


# 4. Save matches as TXT
def save_matches_to_txt(match_records):

    try:

        with open(TXT_FILE, "w") as file:

            file.write("LA LIGA FOOTBALL MATCH RECORDS\n")
            file.write("=" * 50 + "\n\n")

            for match in match_records:

                file.write(
                    f"Match ID: {match['match_id']}\n"
                )

                file.write(
                    f"Date: {match['date']}\n"
                )

                file.write(
                    f"Home Team: {match['home_team']}\n"
                )

                file.write(
                    f"Away Team: {match['away_team']}\n"
                )

                file.write(
                    f"Score: {match['home_goals']} - "
                    f"{match['away_goals']}\n"
                )

                file.write(
                    f"Result: {match['result']}\n"
                )

                file.write("-" * 50 + "\n")

        print("TXT file saved successfully!")

    except (OSError, KeyError) as error:

        print("TXT Save Error:", error)

# TEST FILE HANDLING

print("FILE MANAGEMENT TEST")

# Save all match records
save_matches_to_json(matches)
save_matches_to_csv(matches)
save_matches_to_txt(matches)

print("\nLoading data from JSON file...")

loaded_matches = load_matches_from_json()

if loaded_matches is not None:

    # Update existing match list
    matches[:] = loaded_matches

    # Update DataFrame
    matches_df = pd.DataFrame(matches)

    print("Total Loaded Matches:", len(matches))
    print("Match data loaded successfully!")

print("\nProject files created:")
print(JSON_FILE)
print(CSV_FILE)
print(TXT_FILE)
```

```python

# PART 10: TEAM STATISTICS AND LEAGUE TABLE


# CALCULATE TEAM STATISTICS

def calculate_team_statistics(match_records):

    teams = set()

    for match in match_records:
        teams.add(match["home_team"].strip())
        teams.add(match["away_team"].strip())

    statistics = {}

    for team in teams:
        statistics[team] = {
            "Played": 0,
            "Won": 0,
            "Drawn": 0,
            "Lost": 0,
            "GF": 0,
            "GA": 0,
            "GD": 0,
            "Points": 0
        }

    for match in match_records:

        home = match["home_team"].strip()
        away = match["away_team"].strip()

        hg = int(match["home_goals"])
        ag = int(match["away_goals"])

        statistics[home]["Played"] += 1
        statistics[away]["Played"] += 1

        statistics[home]["GF"] += hg
        statistics[home]["GA"] += ag

        statistics[away]["GF"] += ag
        statistics[away]["GA"] += hg

        if hg > ag:

            statistics[home]["Won"] += 1
            statistics[away]["Lost"] += 1

            statistics[home]["Points"] += 3

        elif hg < ag:

            statistics[away]["Won"] += 1
            statistics[home]["Lost"] += 1

            statistics[away]["Points"] += 3

        else:

            statistics[home]["Drawn"] += 1
            statistics[away]["Drawn"] += 1

            statistics[home]["Points"] += 1
            statistics[away]["Points"] += 1

    # Calculate Goal Difference

    for team in statistics:

        statistics[team]["GD"] = (
            statistics[team]["GF"] -
            statistics[team]["GA"]
        )

    return statistics


# SORTING FUNCTION (WITHOUT LAMBDA)

def sort_by_points(item):

    return (
        item["Points"],
        item["Goal Difference"]
    )


# CREATE LEAGUE TABLE

def create_league_table(match_records):

    statistics = calculate_team_statistics(match_records)

    table = []

    for team, stats in statistics.items():

        row = {
            "Team": team,
            "Played": stats["Played"],
            "Wins": stats["Won"],
            "Draws": stats["Drawn"],
            "Losses": stats["Lost"],
            "Goals For": stats["GF"],
            "Goals Against": stats["GA"],
            "Goal Difference": stats["GD"],
            "Points": stats["Points"]
        }

        table.append(row)

    # Sort by Points and Goal Difference

    table.sort(
        key=sort_by_points,
        reverse=True
    )

    return pd.DataFrame(table)


# DISPLAY TEAM STATISTICS

print("LA LIGA TEAM STATISTICS")

team_statistics = calculate_team_statistics(matches)

if matches:

    for team, stats in team_statistics.items():

        print("\nTeam:", team)
        print("Played:", stats["Played"])
        print("Wins:", stats["Won"])
        print("Draws:", stats["Drawn"])
        print("Losses:", stats["Lost"])
        print("Goals For:", stats["GF"])
        print("Goals Against:", stats["GA"])
        print("Goal Difference:", stats["GD"])
        print("Points:", stats["Points"])

else:

    print("No match records available.")


# DISPLAY LEAGUE TABLE

league_table = create_league_table(matches)

print("\nLEAGUE TABLE")

print(league_table.to_string(index=False))


# NUMPY CALCULATION: AVERAGE GOALS

if matches:

    total_goals = []

    for match in matches:

        goals = (
            int(match["home_goals"]) +
            int(match["away_goals"])
        )

        total_goals.append(goals)

    average_goals = np.mean(total_goals)

else:

    average_goals = 0


# DISPLAY SUMMARY

print("\nLEAGUE SUMMARY")

print("Total Teams:", len(team_statistics))
print("Total Matches:", len(matches))
print("Average Goals Per Match:", round(average_goals, 2))
```

```python
# PART 11: INTERACTIVE MATCH PREDICTION


# PREDICT MATCH FUNCTION

def predict_match(home_team, away_team):

    home_team = home_team.strip()
    away_team = away_team.strip()

    if not home_team or not away_team:
        raise ValueError("Both team names are required.")

    if home_team.lower() == away_team.lower():
        raise ValueError("Home and Away teams cannot be the same.")

    statistics = calculate_team_statistics(matches)

    if home_team not in statistics:
        raise ValueError(
            home_team + " has no match records. Please add a match first."
        )

    if away_team not in statistics:
        raise ValueError(
            away_team + " has no match records. Please add a match first."
        )

    home_stats = statistics[home_team]
    away_stats = statistics[away_team]

    home_average = (
        home_stats["Points"] / home_stats["Played"]
        if home_stats["Played"] > 0 else 0
    )

    away_average = (
        away_stats["Points"] / away_stats["Played"]
        if away_stats["Played"] > 0 else 0
    )

    # Home advantage

    home_score = home_average + 0.5
    away_score = away_average

    if home_score > away_score:
        prediction = "Home Win"

    elif away_score > home_score:
        prediction = "Away Win"

    else:
        prediction = "Draw"

    return prediction, home_average, away_average


# TAKE USER INPUT

print("LA LIGA MATCH PREDICTOR")

print("\nAvailable Teams:")

# Get teams from existing match records

available_teams = set()

for match in matches:
    available_teams.add(match["home_team"].strip())
    available_teams.add(match["away_team"].strip())

if len(available_teams) < 2:

    print("At least two teams with match records are required.")

else:

    for team in sorted(available_teams):
        print("-", team)

    try:

        home_team = input("\nEnter Home Team: ").strip()
        away_team = input("Enter Away Team: ").strip()

        prediction, home_avg, away_avg = predict_match(
            home_team, away_team
        )

        print("\n========== PREDICTION RESULT ==========")

        print("Home Team:", home_team)
        print("Away Team:", away_team)

        print("Home Average Points:", round(home_avg, 2))
        print("Away Average Points:", round(away_avg, 2))

        print("Predicted Result:", prediction)

    except ValueError as error:

        print("\nInput Error:", error)

print("\nPrediction process completed.")
```

```python
# LA LIGA FOOTBALL MATCH PREDICTION
# COMPLETE 7-PAGE INTERACTIVE DASHBOARD



# 1. FILE SETTINGS

DATA_FOLDER = "la_liga_data"
os.makedirs(DATA_FOLDER, exist_ok=True)

JSON_FILE = os.path.join(DATA_FOLDER, "matches.json")
CSV_FILE = os.path.join(DATA_FOLDER, "matches.csv")
TXT_FILE = os.path.join(DATA_FOLDER, "matches.txt")

# 2. COMMON FUNCTIONS

def prepare_all_matches():

    global matches_df

    for match in matches:

        home_goals = int(match["home_goals"])
        away_goals = int(match["away_goals"])

        if home_goals > away_goals:
            match["result"] = "Home Win"
            match["home_points"] = 3
            match["away_points"] = 0

        elif home_goals < away_goals:
            match["result"] = "Away Win"
            match["home_points"] = 0
            match["away_points"] = 3

        else:
            match["result"] = "Draw"
            match["home_points"] = 1
            match["away_points"] = 1

    matches_df = pd.DataFrame(matches)


def save_all_match_files():

    prepare_all_matches()

    with open(JSON_FILE, "w", encoding="utf-8") as file:
        json.dump(matches, file, indent=4, ensure_ascii=False)

    if matches:

        fieldnames = list(matches[0].keys())

        with open(CSV_FILE, "w", newline="", encoding="utf-8") as file:
            writer = csv.DictWriter(file, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(matches)

    else:

        with open(CSV_FILE, "w", newline="", encoding="utf-8") as file:
            file.write("")

    with open(TXT_FILE, "w", encoding="utf-8") as file:

        for match in matches:

            file.write(
                f"Match ID: {match['match_id']}\n"
                f"Date: {match['date']}\n"
                f"Home Team: {match['home_team']}\n"
                f"Away Team: {match['away_team']}\n"
                f"Score: {match['home_goals']} - {match['away_goals']}\n"
                f"Result: {match['result']}\n"
                f"{'-' * 35}\n"
            )


def load_match_data():

    global matches

    if not os.path.exists(JSON_FILE):
        raise FileNotFoundError(
            "No saved JSON file found. Save records first."
        )

    with open(JSON_FILE, "r", encoding="utf-8") as file:
        loaded_data = json.load(file)

    if not isinstance(loaded_data, list):
        raise ValueError("Invalid JSON data format.")

    required_fields = [
        "match_id", "date", "home_team",
        "away_team", "home_goals", "away_goals"
    ]

    checked_records = []
    seen_ids = set()

    for record in loaded_data:

        if not isinstance(record, dict):
            raise ValueError("Invalid match record found.")

        for field in required_fields:
            if field not in record:
                raise ValueError("A required field is missing.")

        match_id = str(record["match_id"]).strip().upper()

        if not match_id or match_id in seen_ids:
            raise ValueError("Empty or duplicate Match ID found.")

        if record["home_team"] not in TEAM_NAMES:
            raise ValueError("Invalid home team found.")

        if record["away_team"] not in TEAM_NAMES:
            raise ValueError("Invalid away team found.")

        if record["home_team"] == record["away_team"]:
            raise ValueError(
                "Home and away teams cannot be the same."
            )

        home_goals = int(record["home_goals"])
        away_goals = int(record["away_goals"])

        if not (0 <= home_goals <= 20 and 0 <= away_goals <= 20):
            raise ValueError("Goals must be between 0 and 20.")

        datetime.strptime(str(record["date"]), "%Y-%m-%d")

        clean_record = {
            "match_id": match_id,
            "date": str(record["date"]),
            "home_team": record["home_team"],
            "away_team": record["away_team"],
            "home_goals": home_goals,
            "away_goals": away_goals
        }

        checked_records.append(clean_record)
        seen_ids.add(match_id)

    matches.clear()
    matches.extend(checked_records)

    prepare_all_matches()


def refresh_dashboard():

    prepare_all_matches()

    with dashboard_output:

        clear_output(wait=True)

        total_matches = len(matches)

        total_goals = sum(
            match["home_goals"] + match["away_goals"]
            for match in matches
        )

        total_draws = sum(
            1 for match in matches
            if match["result"] == "Draw"
        )

        print("LA LIGA FOOTBALL DASHBOARD")
        print("Total Matches:", total_matches)
        print("Total Teams:", len(TEAM_NAMES))
        print("Total Goals:", total_goals)
        print("Total Draws:", total_draws)

        print("\nLEAGUE TABLE")

        if matches:
            display(create_league_table(matches))
        else:
            print("No match records available.")


def get_team_statistics(team_name):

    played = 0
    wins = 0
    draws = 0
    losses = 0
    goals_for = 0
    goals_against = 0
    points = 0

    for match in matches:

        if match["home_team"] == team_name:

            played += 1
            goals_for += match["home_goals"]
            goals_against += match["away_goals"]

            if match["home_goals"] > match["away_goals"]:
                wins += 1
                points += 3

            elif match["home_goals"] == match["away_goals"]:
                draws += 1
                points += 1

            else:
                losses += 1

        elif match["away_team"] == team_name:

            played += 1
            goals_for += match["away_goals"]
            goals_against += match["home_goals"]

            if match["away_goals"] > match["home_goals"]:
                wins += 1
                points += 3

            elif match["away_goals"] == match["home_goals"]:
                draws += 1
                points += 1

            else:
                losses += 1

    return {
        "Team": team_name,
        "Played": played,
        "Wins": wins,
        "Draws": draws,
        "Losses": losses,
        "Goals For": goals_for,
        "Goals Against": goals_against,
        "Goal Difference": goals_for - goals_against,
        "Points": points
    }


# 3. PAGE 1: DASHBOARD

dashboard_output = widgets.Output()

dashboard_page = widgets.VBox([
    widgets.HTML("<h2>La Liga Football Dashboard</h2>"),
    widgets.HTML(
        "<p>Overview of match records and league performance.</p>"
    ),
    dashboard_output
])


# 4. PAGE 2: ADD MATCH

add_id = widgets.Text(
    description="Match ID:",
    placeholder="Example: M013"
)

add_date = widgets.Text(
    description="Date:",
    placeholder="YYYY-MM-DD"
)

add_home = widgets.Text(
    description="Home Team:",
    placeholder="Enter home club name"
)

add_away = widgets.Text(
    description="Away Team:",
    placeholder="Enter away club name"
)

add_home_goals = widgets.BoundedIntText(
    description="Home Goals:",
    min=0,
    max=20,
    value=0
)

add_away_goals = widgets.BoundedIntText(
    description="Away Goals:",
    min=0,
    max=20,
    value=0
)

add_button = widgets.Button(
    description="Add Match",
    button_style="success",
    icon="plus"
)

add_output = widgets.Output()


def on_add_match(button):

    with add_output:

        clear_output(wait=True)

        try:

            match_id = add_id.value.strip().upper()
            match_date = add_date.value.strip()

            if not match_id:
                raise ValueError("Match ID cannot be empty.")

            if any(
                match["match_id"] == match_id
                for match in matches
            ):
                raise ValueError("This Match ID already exists.")

            datetime.strptime(match_date, "%Y-%m-%d")

            if add_home.value == add_away.value:
                raise ValueError(
                    "Home and away teams must be different."
                )

            new_match = {
                "match_id": match_id,
                "date": match_date,
                "home_team": add_home.value,
                "away_team": add_away.value,
                "home_goals": add_home_goals.value,
                "away_goals": add_away_goals.value
            }

            matches.append(new_match)
            update_stats_team_dropdown()
            update_predictor_dropdowns()

            prepare_all_matches()
            save_all_match_files()
            refresh_dashboard()

            print("Match added successfully!\n")
            print("Match ID:", match_id)
            print("Date:", match_date)
            print("Home Team:", add_home.value)
            print("Away Team:", add_away.value)
            print("Home Goals:", add_home_goals.value)
            print("Away Goals:", add_away_goals.value)

            add_id.value = ""
            add_date.value = ""
            add_home_goals.value = 0
            add_away_goals.value = 0

            add_id.value = ""
            add_date.value = ""
            add_home_goals.value = 0
            add_away_goals.value = 0

        except ValueError as error:
            print("Error:", error)

        except (OSError, TypeError) as error:
            print("Data or File Error:", error)

        except Exception as error:
            print("Unexpected Error:", error)


add_button.on_click(on_add_match)

add_page = widgets.VBox([
    widgets.HTML("<h2>Add New Match</h2>"),
    add_id,
    add_date,
    add_home,
    add_away,
    add_home_goals,
    add_away_goals,
    add_button,
    add_output
])

# 5. PAGE 3: MATCH RECORDS


search_input = widgets.Text(
    description="Search:",
    placeholder="Enter Match ID or Team Name"
)

search_button = widgets.Button(
    description="Search Match",
    button_style="info",
    icon="search"
)

show_all_button = widgets.Button(
    description="Show All Matches",
    button_style="primary"
)

records_output = widgets.Output()


def on_search_match(button):

    with records_output:

        clear_output(wait=True)

        search_text = search_input.value.strip().lower()

        if not search_text:
            print("Please enter a Match ID or Team Name.")
            return

        found_records = []

        for match in matches:

            if (
                search_text in match["match_id"].lower()
                or search_text in match["home_team"].lower()
                or search_text in match["away_team"].lower()
            ):
                found_records.append(match)

        if found_records:
            display(pd.DataFrame(found_records))
        else:
            print("No matching records found.")


def on_show_all_matches(button):

    with records_output:

        clear_output(wait=True)

        if matches:
            prepare_all_matches()
            display(pd.DataFrame(matches))
        else:
            print("No match records available.")


search_button.on_click(on_search_match)
show_all_button.on_click(on_show_all_matches)

records_page = widgets.VBox([
    widgets.HTML("<h2>Match Records</h2>"),
    search_input,
    widgets.HBox([search_button, show_all_button]),
    records_output
])

# 6. PAGE 4: UPDATE / DELETE

edit_id = widgets.Text(
    description="Match ID:",
    placeholder="Enter Match ID"
)

find_edit_button = widgets.Button(
    description="Find Match",
    button_style="info"
)

edit_home_goals = widgets.BoundedIntText(
    description="Home Goals:",
    min=0,
    max=20,
    value=0
)

edit_away_goals = widgets.BoundedIntText(
    description="Away Goals:",
    min=0,
    max=20,
    value=0
)

update_button = widgets.Button(
    description="Update Match",
    button_style="success"
)

delete_button = widgets.Button(
    description="Delete Match",
    button_style="danger"
)

edit_output = widgets.Output()


def on_find_edit(button):

    with edit_output:

        clear_output(wait=True)

        match_id = edit_id.value.strip().upper()

        if not match_id:
            print("Please enter a Match ID.")
            return

        for match in matches:

            if match["match_id"] == match_id:

                edit_home_goals.value = int(match["home_goals"])
                edit_away_goals.value = int(match["away_goals"])

                print("Match found!")
                display(pd.DataFrame([match]))
                return

        print("Match not found.")


def on_update_match(button):

    with edit_output:

        clear_output(wait=True)

        match_id = edit_id.value.strip().upper()

        if not match_id:
            print("Please enter a Match ID.")
            return

        for match in matches:

            if match["match_id"] == match_id:

                old_home_goals = match["home_goals"]
                old_away_goals = match["away_goals"]

                try:

                    match["home_goals"] = edit_home_goals.value
                    match["away_goals"] = edit_away_goals.value

                    prepare_all_matches()
                    save_all_match_files()
                    refresh_dashboard()

                    print("Match updated successfully!")
                    return

                except Exception as error:

                    match["home_goals"] = old_home_goals
                    match["away_goals"] = old_away_goals
                    prepare_all_matches()

                    print("Update Error:", error)
                    return

        print("Match not found. Please check the Match ID.")


def on_delete_match(button):

    with edit_output:

        clear_output(wait=True)

        match_id = edit_id.value.strip().upper()

        if not match_id:
            print("Please enter a Match ID.")
            return

        for match in matches:

            if match["match_id"] == match_id:

                matches.remove(match)

                try:

                    prepare_all_matches()
                    save_all_match_files()
                    refresh_dashboard()

                    print("Match deleted successfully!")

                except Exception as error:
                    print("Delete/File Error:", error)

                return

        print("Match not found. Please check the Match ID.")


find_edit_button.on_click(on_find_edit)
update_button.on_click(on_update_match)
delete_button.on_click(on_delete_match)

edit_page = widgets.VBox([
    widgets.HTML("<h2>Update / Delete Match</h2>"),
    edit_id,
    find_edit_button,
    edit_home_goals,
    edit_away_goals,
    widgets.HBox([update_button, delete_button]),
    edit_output
])

# 7. PAGE 5: TEAM STATISTICS

def get_available_teams():

    available_teams = set()

    for match in matches:

        available_teams.add(match["home_team"].strip())
        available_teams.add(match["away_team"].strip())

    return sorted(available_teams)


# Team dropdown

stats_team = widgets.Dropdown(
    options=get_available_teams(),
    description="Select Team:"
)


# Update dropdown whenever new teams are added

def update_stats_team_dropdown():

    teams = get_available_teams()

    old_team = stats_team.value

    stats_team.options = teams

    if old_team in teams:
        stats_team.value = old_team

    elif len(teams) > 0:
        stats_team.value = teams[0]


stats_button = widgets.Button(
    description="Show Statistics",
    button_style="info"
)

stats_table_button = widgets.Button(
    description="Show League Table",
    button_style="success"
)

stats_output = widgets.Output()


# Show selected team statistics

def on_show_statistics(button):

    with stats_output:

        clear_output(wait=True)

        update_stats_team_dropdown()

        selected_team = stats_team.value

        if not selected_team:
            print("No team records available.")
            return

        statistics = get_team_statistics(selected_team)

        print("=" * 45)
        print("             TEAM STATISTICS")
        print("=" * 45)

        for key, value in statistics.items():

            print(key + ":", value)

        if statistics["Played"] > 0:

            average_points = (
                statistics["Points"] / statistics["Played"]
            )

            print(
                "Average Points per Match:",
                round(average_points, 2)
            )

        print("=" * 45)


# Show complete league table

def on_show_league_table(button):

    with stats_output:

        clear_output(wait=True)

        if len(matches) == 0:
            print("No match records available.")
            return

        league_table = create_league_table(matches)

        print("             LA LIGA LEAGUE TABLE")
        print("=" * 100)

        print(league_table.to_string(index=False))

        print("=" * 100)


# Button connections

stats_button.on_click(on_show_statistics)

stats_table_button.on_click(on_show_league_table)


# Statistics page

stats_page = widgets.VBox([

    widgets.HTML("<h2>Team Statistics</h2>"),

    stats_team,

    widgets.HBox([
        stats_button,
        stats_table_button
    ]),

    stats_output

])

# 8. PAGE 6: MATCH PREDICTOR

def get_predictor_teams():

    teams = set()

    for match in matches:
        home = match["home_team"].strip()
        away = match["away_team"].strip()

        if home:
            teams.add(home)

        if away:
            teams.add(away)

    return sorted(teams)


predict_home = widgets.Dropdown(
    options=get_predictor_teams(),
    description="Home Team:"
)

predict_away = widgets.Dropdown(
    options=get_predictor_teams(),
    description="Away Team:"
)

predict_button = widgets.Button(
    description="Predict Match",
    button_style="success",
    icon="futbol"
)

predict_output = widgets.Output()


def update_predictor_dropdowns():

    teams = get_predictor_teams()

    old_home = predict_home.value
    old_away = predict_away.value

    predict_home.options = teams
    predict_away.options = teams

    if len(teams) >= 2:

        if old_home in teams:
            predict_home.value = old_home
        else:
            predict_home.value = teams[0]

        if old_away in teams and old_away != predict_home.value:
            predict_away.value = old_away
        else:
            for team in teams:
                if team != predict_home.value:
                    predict_away.value = team
                    break

    elif len(teams) == 1:
        predict_home.value = teams[0]
        predict_away.value = teams[0]


def on_predict_match(button):

    with predict_output:
        clear_output(wait=True)

        try:

            update_predictor_dropdowns()

            home_team = predict_home.value
            away_team = predict_away.value

            if not home_team or not away_team:
                print("No teams available. Please add match records first.")
                return

            if home_team == away_team:
                print("Please select two different teams.")
                return

            home_stats = get_team_statistics(home_team)
            away_stats = get_team_statistics(away_team)

            if home_stats["Played"] == 0:
                print("No match records found for", home_team)
                return

            if away_stats["Played"] == 0:
                print("No match records found for", away_team)
                return

            home_average = (
                home_stats["Points"] / home_stats["Played"]
            )

            away_average = (
                away_stats["Points"] / away_stats["Played"]
            )

            home_score = home_average + 0.5
            away_score = away_average

            if home_score > away_score:
                prediction = "Home Win"

            elif away_score > home_score:
                prediction = "Away Win"

            else:
                prediction = "Draw"

            print("=" * 45)
            print("             MATCH PREDICTION")
            print("=" * 45)

            print("Home Team:", home_team)
            print("Away Team:", away_team)

            print("-" * 45)

            print("Home Average Points:", round(home_average, 2))
            print("Away Average Points:", round(away_average, 2))

            print("-" * 45)

            print("Predicted Outcome:", prediction)

            print("=" * 45)
            print("Note: This is a simple educational prediction.")
            print("It is not a guaranteed match result.")

        except Exception as error:
            print("Prediction Error:", error)


predict_button.on_click(on_predict_match)

predict_page = widgets.VBox([
    widgets.HTML("<h2>Match Predictor</h2>"),
    predict_home,
    predict_away,
    predict_button,
    predict_output
])


# 9. PAGE 7: DATA MANAGEMENT

save_load_button = widgets.Button(
    description="Save and Load",
    button_style="success",
    icon="save"
)

view_records_button = widgets.Button(
    description="View Match Records",
    button_style="primary",
    icon="table"
)

clear_button = widgets.Button(
    description="Clear",
    button_style="warning",
    icon="eraser"
)

back_dashboard_button = widgets.Button(
    description="Back to Dashboard",
    button_style="info",
    icon="home"
)

dm_output = widgets.Output()


def on_save_load(button):

    with dm_output:

        clear_output(wait=True)

        try:

            save_all_match_files()

            print("All match records saved successfully!")

            print("\nSaved files:")
            print(JSON_FILE)
            print(CSV_FILE)
            print(TXT_FILE)

            load_match_data()

            print("\nMatch records loaded successfully!")
            print("Total Loaded Records:", len(matches))

        except FileNotFoundError as error:
            print("File Error:", error)

        except json.JSONDecodeError:
            print("Error: JSON file is empty or incorrectly formatted.")

        except (ValueError, TypeError) as error:
            print("Data Error:", error)

        except Exception as error:
            print("Unexpected Error:", error)


def on_view_match_records(button):

    with dm_output:

        clear_output(wait=True)

        if matches:
            prepare_all_matches()
            display(pd.DataFrame(matches))
        else:
            print("No match records available.")


# CLEAR FUNCTION

def on_clear_data_output(button):

    with dm_output:
        clear_output(wait=True)

        print("Output cleared successfully.")


def on_back_to_dashboard(button):

    dashboard.selected_index = 0


save_load_button.on_click(on_save_load)
view_records_button.on_click(on_view_match_records)
clear_button.on_click(on_clear_data_output)
back_dashboard_button.on_click(on_back_to_dashboard)


data_management_page = widgets.VBox([

    widgets.HTML("<h2>Data Management</h2>"),

    widgets.HTML(
        "<p>Save, load and view match records.</p>"
    ),

    save_load_button,
    view_records_button,
    clear_button,
    back_dashboard_button,
    dm_output
])

# 10. CREATE 7 DASHBOARD TABS

page_names = [
    "Dashboard",
    "Add Match",
    "Match Records",
    "Update / Delete",
    "Team Statistics",
    "Match Predictor",
    "Data Management"
]

page_contents = [
    dashboard_page,
    add_page,
    records_page,
    edit_page,
    stats_page,
    predict_page,
    data_management_page
]

dashboard = widgets.Tab(children=page_contents)

for i, name in enumerate(page_names):
    dashboard.set_title(i, name)

# 11. DISPLAY DASHBOARD

prepare_all_matches()
refresh_dashboard()

print("La Liga Football Dashboard is ready!")

display(dashboard)
```
